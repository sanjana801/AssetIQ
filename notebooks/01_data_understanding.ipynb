```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
file_path = r"D:\AssetIQ\data\raw\cmapss\train_FD001.txt"

df = pd.read_csv(file_path, sep=r"\s+", header=None)

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
columns = (
    ["unit", "cycle", "setting_1", "setting_2", "setting_3"]
    + [f"sensor_{i}" for i in range(1, 22)]
)

df.columns = columns

df.head()
```

```python
df.info()
```

```python
sensor_cols = [f"sensor_{i}" for i in range(1, 22)]

df[sensor_cols].nunique().sort_values()
```

```python
constant_sensors = [col for col in sensor_cols if df[col].nunique() == 1]

df[constant_sensors].iloc[0]
```

```python
constant_sensors = [col for col in sensor_cols if df[col].nunique() == 1]

df_model = df.drop(columns=constant_sensors).copy()

print("Removed:", constant_sensors)
print("Original columns:", df.shape[1])
print("Remaining columns:", df_model.shape[1])
```

```python
engine_life = df.groupby("unit")["cycle"].max()

print("Number of engines:", engine_life.shape[0])
print("Minimum lifetime:", engine_life.min())
print("Maximum lifetime:", engine_life.max())
print("Average lifetime:", engine_life.mean())
```

```python
engine_life.describe()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.bar(engine_life.index, engine_life.values)
plt.xlabel("Engine Unit")
plt.ylabel("Total Operating Cycles")
plt.title("FD001 Engine Lifetimes")
plt.show()
```
